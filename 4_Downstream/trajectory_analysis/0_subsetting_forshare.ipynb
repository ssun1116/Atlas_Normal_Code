```python
import scanpy as sc
import pandas as pd
import os
import numpy as np
```

# Calling data, copying matrix

```python
adata = sc.read_h5ad("path to adata")
```

```python
adata.layers['counts'] = adata.layers['counts'].copy()
```

# Subsetting

# RG, Neuroblast, Neurons under 20

```python
target = ['RG', 'Neuroblast', 'Ext', 'Inh']
```

```python
cutoff_age = ["Fetal (1st trimester)", "Fetal (2nd trimester)", "Fetal (3rd trimester)", "Neonatal", "Childhood (1-6Y)", "Childhood (6-12Y)", "Adolescence (12-20Y)"]
```

```python
neuron_under_20 = adata[(adata.obs["Stage"].isin(cutoff_age)) & (adata.obs["Cell Type"].isin(target))]
```

```python
neuron_under_20.write_h5ad("/data1/trajectory_JL/step_1_subsetted_data/neuron_under_20.h5ad")
```

# ODC_lin

```python
odc = adata[adata.obs["Cell Type"].isin(["OPC", "Oligo"])].copy()
```

```python
odc.write_h5ad("/data1/trajectory_JL/step_1_subsetted_data/odc.h5ad")
```

# Astro

```python
astro = adata[adata.obs["Cell Type"] == "Astro"].copy()
```

```python
astro.write_h5ad("/data1/trajectory_JL/step_1_subsetted_data/astro.h5ad")
```
